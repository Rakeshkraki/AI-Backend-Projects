```python
#porterstem
my_words = ["eat", "eats", "eating", "give", "gives", "giving", "tell", "tells", "telling"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in my_words:
    print(f" {word} -------------> {stemming.stem(word)}")
```

```python
from nltk.stem import RegexpStemmer
```

```python
regex_stemmer = RegexpStemmer(regexp= 'ing$|s$|e$',min = 4)
```

```python
regex_stemmer.stem('eats')
```

```python
from nltk.stem import SnowballStemmer
snow_ball_stemmer = SnowballStemmer("english")
```

```python
for word in my_words:
    print(f" {word} -------------> {snow_ball_stemmer.stem(word)}")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
for word in my_words:
    print(f" {word} -------------> {lemmatizer.lemmatize(word, pos='v')}")
```

```python
""" TF-IDF is a statistical technique that evaluates how important a word is in a document relative to a collection of documents. It reduces the weight of common words and increases the weight of rare words."""
```

```python
from gensim.models import Word2Vec

sentences = [
    ["I", "love", "AI"],
    ["I", "love", "programming"],
    ["AI", "is", "powerful"]
]

model = Word2Vec(sentences, vector_size=50, window=2, min_count=1)

print(model.wv["AI"])
print(model.wv.most_similar("AI"))

```
